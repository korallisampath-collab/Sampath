```python

# ============================================================
# STEP 2: MENTAL HEALTH SAFETY AND INPUT PROCESSING
# ============================================================

from pathlib import Path
import re
import json

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

# Create safety module directory
SAFETY_DIR = PROJECT_DIR / "src" / "safety"
SAFETY_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Process user input
# ------------------------------------------------------------

def process_user_input(message):
    """Validate and normalize a user's text message."""

    if not isinstance(message, str):
        return {
            "valid": False,
            "message": "",
            "error": "Input must be text."
        }

    message = re.sub(r"\s+", " ", message).strip()

    if not message:
        return {
            "valid": False,
            "message": "",
            "error": "Please enter a message."
        }

    if len(message) > 5000:
        return {
            "valid": False,
            "message": "",
            "error": "Message is too long. Limit: 5000 characters."
        }

    return {
        "valid": True,
        "message": message,
        "error": None
    }


# ------------------------------------------------------------
# 2. Basic safety screening
# ------------------------------------------------------------

CRISIS_PHRASES = [
    "kill myself",
    "suicide",
    "end my life",
    "want to die",
    "hurt myself",
    "self harm",
    "self-harm"
]


def screen_safety(message):
    """
    Flag messages containing basic crisis-related phrases.
    This is a simple demonstration, not a clinical assessment.
    """

    normalized = message.lower()

    matched_phrases = [
        phrase for phrase in CRISIS_PHRASES
        if phrase in normalized
    ]

    if matched_phrases:
        return {
            "safety_flag": True,
            "route": "SAFETY_FIRST",
            "reason": "Potential crisis-related language detected.",
            "matched_phrases": matched_phrases
        }

    return {
        "safety_flag": False,
        "route": "NORMAL_SUPPORT",
        "reason": "No configured crisis phrase detected.",
        "matched_phrases": []
    }


# ------------------------------------------------------------
# 3. Create a reusable input-processing pipeline
# ------------------------------------------------------------

def prepare_message(message):
    """Validate input and run the initial safety screen."""

    processed = process_user_input(message)

    if not processed["valid"]:
        return {
            "valid": False,
            "error": processed["error"]
        }

    safety = screen_safety(processed["message"])

    return {
        "valid": True,
        "processed_message": processed["message"],
        "safety": safety
    }


# ------------------------------------------------------------
# 4. Save the module for later project steps
# ------------------------------------------------------------

module_code = '''import re

CRISIS_PHRASES = [
    "kill myself", "suicide", "end my life",
    "want to die", "hurt myself", "self harm", "self-harm"
]

def process_user_input(message):
    if not isinstance(message, str):
        return {"valid": False, "message": "", "error": "Input must be text."}
    message = re.sub(r"\\\\s+", " ", message).strip()
    if not message:
        return {"valid": False, "message": "", "error": "Please enter a message."}
    if len(message) > 5000:
        return {"valid": False, "message": "", "error": "Message is too long."}
    return {"valid": True, "message": message, "error": None}

def screen_safety(message):
    normalized = message.lower()
    matched = [p for p in CRISIS_PHRASES if p in normalized]
    return {
        "safety_flag": bool(matched),
        "route": "SAFETY_FIRST" if matched else "NORMAL_SUPPORT",
        "reason": "Potential crisis-related language detected." if matched
                  else "No configured crisis phrase detected.",
        "matched_phrases": matched
    }

def prepare_message(message):
    result = process_user_input(message)
    if not result["valid"]:
        return {"valid": False, "error": result["error"]}
    return {
        "valid": True,
        "processed_message": result["message"],
        "safety": screen_safety(result["message"])
    }
'''

module_path = SAFETY_DIR / "input_safety.py"
module_path.write_text(module_code, encoding="utf-8")

# ------------------------------------------------------------
# 5. Test the pipeline
# ------------------------------------------------------------

test_messages = [
    "I am feeling stressed about my exams.",
    "I feel sad and want to talk to someone.",
    "I want to kill myself.",
    "   "
]

print("=" * 60)
print("STEP 2: INPUT AND SAFETY TEST RESULTS")
print("=" * 60)

for number, message in enumerate(test_messages, start=1):
    print(f"\nTest {number}: {message!r}")
    result = prepare_message(message)
    print(json.dumps(result, indent=2))

# ------------------------------------------------------------
# 6. Verify that the module was saved
# ------------------------------------------------------------

assert module_path.exists()

print("\n" + "=" * 60)
print("STEP 2 MODULE SAVED SUCCESSFULLY")
print("File:", module_path)
print("=" * 60)
```

```python

# ============================================================
# PROJECT 1: HUMAN ROLE-MODEL INSPIRED COUNSELLING AGENT
# STEP 5: PERSONALIZED USER PROFILE AND PREFERENCES
# ============================================================

from pathlib import Path
import json

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

PROFILE_DIR = PROJECT_DIR / "data"
PROFILE_DIR.mkdir(parents=True, exist_ok=True)

PROFILE_FILE = PROFILE_DIR / "user_profile.json"

# 1. Define allowed preferences

ALLOWED_STYLES = [
    "empathetic",
    "motivational",
    "practical"
]

ALLOWED_GOALS = [
    "manage_stress",
    "improve_confidence",
    "build_routines",
    "handle_anxiety",
    "general_wellness"
]

ALLOWED_LENGTHS = [
    "short",
    "detailed"
]


# 2. Create and validate a profile

def create_user_profile(
    support_style="empathetic",
    main_goal="general_wellness",
    response_length="short"
):
    """Create a profile using validated preferences."""

    support_style = str(support_style).strip().lower()
    main_goal = str(main_goal).strip().lower()
    response_length = str(response_length).strip().lower()

    if support_style not in ALLOWED_STYLES:
        raise ValueError(
            "Invalid support style. Choose: "
            + ", ".join(ALLOWED_STYLES)
        )

    if main_goal not in ALLOWED_GOALS:
        raise ValueError(
            "Invalid goal. Choose: "
            + ", ".join(ALLOWED_GOALS)
        )

    if response_length not in ALLOWED_LENGTHS:
        raise ValueError(
            "Invalid response length. Choose: "
            + ", ".join(ALLOWED_LENGTHS)
        )

    return {
        "support_style": support_style,
        "main_goal": main_goal,
        "response_length": response_length
    }


# 3. Save a profile

def save_user_profile(profile, file_path=PROFILE_FILE):
    """Save only the supplied preference fields."""

    validated_profile = create_user_profile(
        support_style=profile["support_style"],
        main_goal=profile["main_goal"],
        response_length=profile["response_length"]
    )

    file_path = Path(file_path)
    file_path.parent.mkdir(parents=True, exist_ok=True)

    with open(file_path, "w", encoding="utf-8") as file:
        json.dump(
            validated_profile,
            file,
            indent=4,
            ensure_ascii=False
        )

    return file_path


# 4. Load a saved profile

def load_user_profile(file_path=PROFILE_FILE):
    file_path = Path(file_path)

    if not file_path.exists():
        return create_user_profile()

    with open(file_path, "r", encoding="utf-8") as file:
        saved_data = json.load(file)

    return create_user_profile(
        support_style=saved_data.get(
            "support_style", "empathetic"
        ),
        main_goal=saved_data.get(
            "main_goal", "general_wellness"
        ),
        response_length=saved_data.get(
            "response_length", "short"
        )
    )


# 5. Create a sample profile for testing

sample_profile = create_user_profile(
    support_style="empathetic",
    main_goal="manage_stress",
    response_length="short"
)

saved_path = save_user_profile(sample_profile)
loaded_profile = load_user_profile()

# 6. Display test results

print("=" * 55)
print("STEP 5: USER PROFILE TEST")
print("=" * 55)

print("\nSample profile:")
print(json.dumps(loaded_profile, indent=4))

assert loaded_profile == sample_profile
assert saved_path.exists()

print("\nProfile validation: PASSED")
print("Profile saving: PASSED")
print("Profile loading: PASSED")
print("STEP 5 SUCCESSFULLY COMPLETED")
```

```python

# ============================================================
# STEP 6: COUNSELLING STRATEGY SELECTION
# ============================================================

from pathlib import Path
import json
import re

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

KNOWLEDGE_FILE = (
    PROJECT_DIR
    / "knowledge_base"
    / "role_model_strategies.json"
)

PROFILE_FILE = PROJECT_DIR / "data" / "user_profile.json"
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Verify required files
# ------------------------------------------------------------

if not KNOWLEDGE_FILE.exists():
    raise FileNotFoundError(
        "Knowledge base not found. Complete Step 4 first."
    )

if not PROFILE_FILE.exists():
    raise FileNotFoundError(
        "User profile not found. Complete Step 5 first."
    )

with open(KNOWLEDGE_FILE, "r", encoding="utf-8") as file:
    knowledge_base = json.load(file)

with open(PROFILE_FILE, "r", encoding="utf-8") as file:
    user_profile = json.load(file)

# ------------------------------------------------------------
# 2. Basic concern detection for this standalone step
# ------------------------------------------------------------

CONCERN_KEYWORDS = {
    "stress": [
        "stress", "stressed", "overwhelmed", "pressure"
    ],
    "sadness": [
        "sad", "unhappy", "lonely", "heartbroken"
    ],
    "anxiety": [
        "anxious", "anxiety", "worried", "nervous", "panic"
    ],
    "low_confidence": [
        "not confident", "low confidence", "self-doubt",
        "self doubt", "not good enough"
    ],
    "motivation": [
        "unmotivated", "no motivation", "procrastinating",
        "cannot focus", "can't focus", "lost motivation"
    ],
    "anger": [
        "angry", "frustrated", "irritated", "furious"
    ]
}


def detect_concerns(message):
    text = re.sub(r"\s+", " ", message.lower()).strip()
    detected = []

    for concern, keywords in CONCERN_KEYWORDS.items():
        for keyword in keywords:
            pattern = r"(?<!\w)" + re.escape(keyword) + r"(?!\w)"

            if re.search(pattern, text):
                detected.append(concern)
                break

    return detected


# ------------------------------------------------------------
# 3. Map profile goals to concern categories
# ------------------------------------------------------------

GOAL_MAPPING = {
    "manage_stress": "stress",
    "improve_confidence": "low_confidence",
    "build_routines": "motivation",
    "handle_anxiety": "anxiety",
    "general_wellness": None
}


# ------------------------------------------------------------
# 4. Select a strategy using concern and preferences
# ------------------------------------------------------------

def select_counselling_strategy(message, profile=None):
    if not isinstance(message, str) or not message.strip():
        return {
            "success": False,
            "error": "Please enter a non-empty message."
        }

    if len(message) > 5000:
        return {
            "success": False,
            "error": "Message is too long."
        }

    if profile is None:
        profile = user_profile

    concerns = detect_concerns(message)

    # Prefer an explicitly detected concern.
    primary_concern = concerns[0] if concerns else None

    # Use the user's goal only if no concern was detected.
    if primary_concern is None:
        primary_concern = GOAL_MAPPING.get(
            profile.get("main_goal", "general_wellness")
        )

    if primary_concern is None:
        return {
            "success": True,
            "detected_concerns": concerns,
            "primary_concern": "unclear",
            "selected_strategy": None,
            "message": (
                "I am not sure what kind of support would help. "
                "Could you tell me a little more about what is "
                "on your mind?"
            )
        }

    # Find relevant strategies.
    candidates = [
        item for item in knowledge_base
        if primary_concern in item.get("suitable_concerns", [])
    ]

    if not candidates:
        return {
            "success": True,
            "detected_concerns": concerns,
            "primary_concern": primary_concern,
            "selected_strategy": None,
            "message": (
                "I could not find a matching strategy. "
                "Please share more about what would help you."
            )
        }

    # Match the strategy to the user's preferred support style.
    style = profile.get("support_style", "empathetic")

    preferred_quality = {
        "empathetic": "self_compassion",
        "motivational": "resilience",
        "practical": "problem_solving"
    }.get(style, "self_compassion")

    selected = next(
        (
            item for item in candidates
            if item.get("quality") == preferred_quality
        ),
        candidates[0]
    )

    # Return the selected strategy.
    return {
        "success": True,
        "detected_concerns": concerns,
        "primary_concern": primary_concern,
        "support_style": style,
        "selected_strategy": {
            "id": selected["id"],
            "name": selected["name"],
            "quality": selected["quality"],
            "strategies": selected["strategies"],
            "exercise": selected["exercise"]
        }
    }


# ------------------------------------------------------------
# 5. Test the strategy selector
# ------------------------------------------------------------

test_messages = [
    "I feel stressed about my exams.",
    "I am not confident in my abilities.",
    "I have no motivation to study.",
    "Hello, I would like some support."
]

print("=" * 60)
print("STEP 6: STRATEGY SELECTION TESTS")
print("=" * 60)

for index, message in enumerate(test_messages, start=1):
    result = select_counselling_strategy(message)

    print(f"\nTest {index}: {message}")
    print("Primary concern:", result.get("primary_concern"))

    selected = result.get("selected_strategy")

    if selected:
        print("Selected strategy:", selected["name"])
        print("Quality:", selected["quality"])
        print("Exercise:", selected["exercise"])
    else:
        print("Response:", result.get("message"))

# ------------------------------------------------------------
# 6. Save a sample result
# ------------------------------------------------------------

sample_result = select_counselling_strategy(
    "I feel stressed about my exams."
)

output_file = OUTPUT_DIR / "strategy_selection_result.json"

with open(output_file, "w", encoding="utf-8") as file:
    json.dump(sample_result, file, indent=4, ensure_ascii=False)

assert sample_result["success"] is True
assert output_file.exists()

print("\nResult saved to:", output_file)
print("STEP 6 SUCCESSFULLY COMPLETED")
```

```python

# ============================================================
# PROJECT 1: HUMAN ROLE-MODEL INSPIRED AI AGENT
# STEP 7: BUILD THE MAIN AI AGENT
# Language: Python
# ============================================================

from pathlib import Path
import json
import re

# ------------------------------------------------------------
# 1. Project paths
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

KNOWLEDGE_FILE = (
    PROJECT_DIR
    / "knowledge_base"
    / "role_model_strategies.json"
)

PROFILE_FILE = PROJECT_DIR / "data" / "user_profile.json"
OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. Check required files
# ------------------------------------------------------------

required_files = [KNOWLEDGE_FILE, PROFILE_FILE]

for file_path in required_files:
    if not file_path.exists():
        raise FileNotFoundError(
            f"Required file missing: {file_path}. "
            "Please complete Steps 4 and 5 first."
        )

with open(KNOWLEDGE_FILE, "r", encoding="utf-8") as file:
    knowledge_base = json.load(file)

with open(PROFILE_FILE, "r", encoding="utf-8") as file:
    user_profile = json.load(file)

assert isinstance(knowledge_base, list)
assert len(knowledge_base) > 0

# ------------------------------------------------------------
# 3. Basic safety screening
# ------------------------------------------------------------

CRISIS_PHRASES = [
    "kill myself",
    "suicide",
    "end my life",
    "want to die",
    "hurt myself",
    "self harm",
    "self-harm"
]


def check_safety(message):
    text = message.lower()

    matches = [
        phrase
        for phrase in CRISIS_PHRASES
        if re.search(
            r"(?<!\w)" + re.escape(phrase) + r"(?!\w)",
            text
        )
    ]

    return {
        "flagged": bool(matches),
        "matched_phrases": matches
    }


# ------------------------------------------------------------
# 4. Detect possible concerns
# ------------------------------------------------------------

CONCERN_KEYWORDS = {
    "stress": [
        "stress", "stressed", "overwhelmed", "pressure"
    ],
    "sadness": [
        "sad", "unhappy", "lonely", "heartbroken"
    ],
    "anxiety": [
        "anxious", "anxiety", "worried", "nervous", "panic"
    ],
    "low_confidence": [
        "not confident", "low confidence",
        "not good enough", "self-doubt", "self doubt"
    ],
    "motivation": [
        "unmotivated", "no motivation",
        "procrastinating", "lost motivation"
    ],
    "anger": [
        "angry", "frustrated", "irritated", "furious"
    ]
}


def detect_concerns(message):
    text = message.lower()
    detected = []

    for concern, keywords in CONCERN_KEYWORDS.items():
        for keyword in keywords:
            pattern = (
                r"(?<!\w)"
                + re.escape(keyword)
                + r"(?!\w)"
            )

            if re.search(pattern, text):
                detected.append(concern)
                break

    return detected


# ------------------------------------------------------------
# 5. Map user goals to concerns
# ------------------------------------------------------------

GOAL_MAPPING = {
    "manage_stress": "stress",
    "improve_confidence": "low_confidence",
    "build_routines": "motivation",
    "handle_anxiety": "anxiety",
    "general_wellness": None
}


# ------------------------------------------------------------
# 6. Select a suitable strategy
# ------------------------------------------------------------

STYLE_MAPPING = {
    "empathetic": "self_compassion",
    "motivational": "resilience",
    "practical": "problem_solving"
}


def choose_strategy(concern, profile):
    if concern is None:
        return None

    candidates = [
        item
        for item in knowledge_base
        if concern in item.get("suitable_concerns", [])
    ]

    if not candidates:
        return None

    preferred_quality = STYLE_MAPPING.get(
        profile.get("support_style", "empathetic"),
        "self_compassion"
    )

    for item in candidates:
        if item.get("quality") == preferred_quality:
            return item

    return candidates[0]


# ------------------------------------------------------------
# 7. Main agent
# ------------------------------------------------------------

def run_agent(message, profile=None):
    """Coordinate safety screening and strategy selection."""

    if not isinstance(message, str):
        return {
            "status": "invalid_input",
            "response": "Please enter your message as text."
        }

    message = message.strip()

    if not message:
        return {
            "status": "invalid_input",
            "response": "Please enter a message so I can help."
        }

    if len(message) > 5000:
        return {
            "status": "invalid_input",
            "response": "Your message is too long. Please shorten it."
        }

    active_profile = (
        user_profile if profile is None else profile
    )

    # Safety screening must happen before normal support.
    safety = check_safety(message)

    if safety["flagged"]:
        return {
            "status": "safety_first",
            "safety_flag": True,
            "detected_concerns": [],
            "selected_strategy": None,
            "response": (
                "I'm sorry you're going through something so difficult. "
                "Your safety matters. If you may act on these thoughts "
                "or are in immediate danger, contact local emergency "
                "services now and try to stay with a trusted person. "
                "You can also contact a qualified mental health "
                "professional. Are you in immediate danger right now?"
            )
        }

    # Detect concerns only after the initial safety check.
    concerns = detect_concerns(message)

    primary_concern = (
        concerns[0] if concerns else None
    )

    if primary_concern is None:
        primary_concern = GOAL_MAPPING.get(
            active_profile.get(
                "main_goal", "general_wellness"
            )
        )

    if primary_concern is None:
        return {
            "status": "needs_clarification",
            "safety_flag": False,
            "detected_concerns": concerns,
            "selected_strategy": None,
            "response": (
                "I'm here to listen. What has been on your mind "
                "recently, and what kind of support would help you?"
            )
        }

    strategy = choose_strategy(
        primary_concern,
        active_profile
    )

    if strategy is None:
        return {
            "status": "needs_clarification",
            "safety_flag": False,
            "detected_concerns": concerns,
            "selected_strategy": None,
            "response": (
                "I'd like to understand your situation better. "
                "Could you tell me what feels most difficult right now?"
            )
        }

    exercise = strategy["exercise"]

    response = (
        "Thank you for sharing that with me. "
        f"It sounds like {primary_concern.replace('_', ' ')} "
        "may be part of what you're experiencing. "
        f"One approach we could explore is {strategy['name']}. "
        f"A small exercise you could try is: {exercise} "
        "Would you like to explore this together?"
    )

    return {
        "status": "support_response",
        "safety_flag": False,
        "detected_concerns": concerns,
        "primary_concern": primary_concern,
        "selected_strategy": strategy["name"],
        "selected_quality": strategy["quality"],
        "suggested_exercise": exercise,
        "response": response
    }


# ------------------------------------------------------------
# 8. Test the main agent
# ------------------------------------------------------------

test_messages = [
    "I am stressed about my exams.",
    "I am not confident in my abilities.",
    "Hello, I need some guidance.",
    "I want to kill myself."
]

print("=" * 65)
print("STEP 7: MAIN AI AGENT TESTING")
print("=" * 65)

for index, message in enumerate(test_messages, start=1):
    result = run_agent(message)

    print(f"\nTEST {index}")
    print("User:", message)
    print("Status:", result["status"])
    print("Agent:", result["response"])

# ------------------------------------------------------------
# 9. Verify and save a demonstration result
# ------------------------------------------------------------

normal_result = run_agent(
    "I am stressed about my exams."
)

safety_result = run_agent(
    "I want to kill myself."
)

assert normal_result["status"] == "support_response"
assert safety_result["status"] == "safety_first"
assert safety_result["selected_strategy"] is None

output_file = OUTPUT_DIR / "main_agent_test_result.json"

with open(output_file, "w", encoding="utf-8") as file:
    json.dump(
        normal_result,
        file,
        indent=4,
        ensure_ascii=False
    )

print("\n" + "=" * 65)
print("Normal-response test: PASSED")
print("Safety-routing test: PASSED")
print("Saved result:", output_file)
print("STEP 7 SUCCESSFULLY COMPLETED")
print("=" * 65)
```

```python

# ============================================================
# PROJECT 1: HUMAN ROLE-MODEL INSPIRED COUNSELLING AGENT
# STEP 9: CONVERSATIONAL MEMORY
# ============================================================

from pathlib import Path
from collections import deque
from datetime import datetime, timezone
import json

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

SRC_DIR = PROJECT_DIR / "src"
OUTPUT_DIR = PROJECT_DIR / "outputs"

SRC_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 1. Create a limited in-memory conversation history
# ------------------------------------------------------------

class ConversationMemory:
    """
    Store a limited number of recent messages in RAM.
    Messages are not automatically written to disk.
    """

    def __init__(self, max_messages=10):
        if not isinstance(max_messages, int) or max_messages < 2:
            raise ValueError(
                "max_messages must be an integer of at least 2."
            )

        self.max_messages = max_messages
        self.messages = deque(maxlen=max_messages)

    def add_message(self, role, content):
        if role not in ("user", "assistant"):
            raise ValueError(
                "Role must be 'user' or 'assistant'."
            )

        if not isinstance(content, str) or not content.strip():
            raise ValueError(
                "Message content must be non-empty text."
            )

        if len(content) > 5000:
            raise ValueError(
                "Message must not exceed 5000 characters."
            )

        entry = {
            "role": role,
            "content": content.strip(),
            "timestamp": datetime.now(
                timezone.utc
            ).isoformat()
        }

        self.messages.append(entry)

    def get_history(self):
        return list(self.messages)

    def clear(self):
        self.messages.clear()

    def count(self):
        return len(self.messages)

    def get_recent_context(self, last_n=4):
        if not isinstance(last_n, int) or last_n < 1:
            raise ValueError(
                "last_n must be a positive integer."
            )

        return self.get_history()[-last_n:]


# ------------------------------------------------------------
# 2. Create a simple conversation manager
# ------------------------------------------------------------

class CounsellingConversation:
    def __init__(self, max_messages=10):
        self.memory = ConversationMemory(max_messages)

    def receive_user_message(self, message):
        self.memory.add_message("user", message)

    def add_agent_response(self, response):
        self.memory.add_message("assistant", response)

    def get_context(self, last_n=4):
        return self.memory.get_recent_context(last_n)

    def reset_conversation(self):
        self.memory.clear()


# ------------------------------------------------------------
# 3. Test the conversation memory
# ------------------------------------------------------------

conversation = CounsellingConversation(max_messages=6)

conversation.receive_user_message(
    "I feel stressed about my exams."
)

conversation.add_agent_response(
    "That sounds difficult. Which subject worries you most?"
)

conversation.receive_user_message(
    "I am worried about mathematics."
)

print("=" * 60)
print("STEP 9: CONVERSATIONAL MEMORY TEST")
print("=" * 60)

print("\nMessages stored:", conversation.memory.count())
print("\nRecent conversation context:")

for item in conversation.get_context(last_n=4):
    print(f"{item['role'].upper()}: {item['content']}")

# Verify bounded memory
for i in range(10):
    conversation.receive_user_message(
        f"Test message {i}"
    )

assert conversation.memory.count() == 6

print("\nMemory limit test: PASSED")
print("Stored messages:", conversation.memory.count())

# Verify reset
conversation.reset_conversation()

assert conversation.memory.count() == 0

print("Conversation reset test: PASSED")
print("Messages after reset:", conversation.memory.count())


# ------------------------------------------------------------
# 4. Save the reusable memory module
# ------------------------------------------------------------

module_source = '''
from collections import deque
from datetime import datetime, timezone


class ConversationMemory:
    def __init__(self, max_messages=10):
        if not isinstance(max_messages, int) or max_messages < 2:
            raise ValueError("max_messages must be an integer of at least 2.")
        self.max_messages = max_messages
        self.messages = deque(maxlen=max_messages)

    def add_message(self, role, content):
        if role not in ("user", "assistant"):
            raise ValueError("Invalid role.")
        if not isinstance(content, str) or not content.strip():
            raise ValueError("Message must be non-empty text.")
        if len(content) > 5000:
            raise ValueError("Message is too long.")

        self.messages.append({
            "role": role,
            "content": content.strip(),
            "timestamp": datetime.now(timezone.utc).isoformat()
        })

    def get_history(self):
        return list(self.messages)

    def get_recent_context(self, last_n=4):
        if not isinstance(last_n, int) or last_n < 1:
            raise ValueError("last_n must be positive.")
        return self.get_history()[-last_n:]

    def clear(self):
        self.messages.clear()

    def count(self):
        return len(self.messages)
'''

module_path = SRC_DIR / "conversation_memory.py"
module_path.write_text(
    module_source.strip() + "\n",
    encoding="utf-8"
)

assert module_path.exists()

print("\nMemory module saved:", module_path)
print("STEP 9 SUCCESSFULLY COMPLETED")
```

```python

# STEP 10.2: LOAD THE LANGUAGE MODEL

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

print("PyTorch:", torch.__version__)

MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Loading model...")
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME)

model.eval()

print("Model loaded successfully!")
print("Device:", next(model.parameters()).device)
print("STEP 10.2 COMPLETED")
```

```python

# ============================================================
# STEP 10.3: GENERATE SUPPORTIVE AI RESPONSES
# ============================================================

import torch

# 1. Verify that the model and tokenizer are loaded
if "model" not in globals() or "tokenizer" not in globals():
    raise RuntimeError(
        "Model or tokenizer missing. Complete Step 10.2 first."
    )


# 2. Define the response-generation function
def generate_supportive_response(user_message):

    if not isinstance(user_message, str):
        return "Please enter your message as text."

    user_message = user_message.strip()

    if not user_message:
        return "Please enter a message."

    if len(user_message) > 3000:
        return "Please shorten your message to 3000 characters or fewer."

    system_prompt = """
You are a supportive mental wellness assistant.

Guidelines:
- Listen empathetically and without judgment.
- Offer practical, gentle, optional wellness exercises.
- Do not diagnose mental health conditions.
- Do not prescribe medication.
- Do not claim to replace a qualified professional.
- Avoid presenting role-model strategies as guaranteed treatments.
- If a person may be in immediate danger, prioritize immediate
  safety and connection to emergency services or trusted people.
- Keep responses concise and ask one helpful follow-up question.
"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_message}
    ]

    # Prepare the prompt for the instruction-tuned model
    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_tensors="pt",
        return_dict=True
    )

    # Move input tensors to the model's device
    inputs = {
        key: value.to(next(model.parameters()).device)
        for key, value in inputs.items()
    }

    # Generate the response
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=120,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )

    # Decode only the newly generated tokens
    input_length = inputs["input_ids"].shape[1]
    new_tokens = outputs[0][input_length:]

    response = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    if not response:
        return (
            "I'm here to listen. Could you tell me more "
            "about what has been troubling you?"
        )

    return response


# 3. Test the function
test_message = "I feel stressed about my upcoming exams."

print("=" * 60)
print("STEP 10.3: SUPPORTIVE RESPONSE GENERATION")
print("=" * 60)
print("User:", test_message)

response = generate_supportive_response(test_message)

print("\nAI response:")
print(response)

assert isinstance(response, str)
assert len(response.strip()) > 0

print("\nSTEP 10.3 COMPLETED SUCCESSFULLY")
```

```python

# ============================================================
# PROJECT 1: HUMAN ROLE-MODEL INSPIRED COUNSELLING AGENT
# STEP 11.1: PERSONALIZED RESPONSE INTEGRATION
# SELF-CONTAINED VERSION
# ============================================================

from pathlib import Path
import json
import re

# ------------------------------------------------------------
# 1. CREATE PROJECT FOLDERS
# ------------------------------------------------------------

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

KNOWLEDGE_DIR = PROJECT_DIR / "knowledge_base"
DATA_DIR = PROJECT_DIR / "data"
OUTPUT_DIR = PROJECT_DIR / "outputs"

for folder in [KNOWLEDGE_DIR, DATA_DIR, OUTPUT_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

KNOWLEDGE_FILE = (
    KNOWLEDGE_DIR / "role_model_strategies.json"
)

PROFILE_FILE = DATA_DIR / "user_profile.json"

OUTPUT_FILE = OUTPUT_DIR / "personalized_agent_result.json"


# ------------------------------------------------------------
# 2. CREATE KNOWLEDGE BASE IF IT IS MISSING
# ------------------------------------------------------------

default_knowledge = [
    {
        "id": "RM001",
        "quality": "resilience",
        "name": "Resilience Approach",
        "suitable_concerns": ["stress", "sadness", "motivation"],
        "exercise": (
            "Identify one challenge and choose one small "
            "action you can take today."
        ),
        "strategies": [
            "Break a difficult task into smaller steps.",
            "Focus on one manageable action."
        ]
    },
    {
        "id": "RM002",
        "quality": "self_confidence",
        "name": "Self-Confidence Approach",
        "suitable_concerns": [
            "low_confidence", "anxiety", "motivation"
        ],
        "exercise": (
            "Write down one personal strength and one "
            "small achievement."
        ),
        "strategies": [
            "Recognize existing strengths.",
            "Set a realistic, achievable goal."
        ]
    },
    {
        "id": "RM003",
        "quality": "mindfulness",
        "name": "Mindfulness Approach",
        "suitable_concerns": ["stress", "anxiety", "anger"],
        "exercise": (
            "Notice five things you can see around you "
            "and take your time with the exercise."
        ),
        "strategies": [
            "Notice your surroundings.",
            "Take comfortable, unforced breaths."
        ]
    },
    {
        "id": "RM004",
        "quality": "discipline",
        "name": "Healthy Routine Approach",
        "suitable_concerns": ["motivation", "stress"],
        "exercise": (
            "Choose one manageable task and work on it "
            "for 15 minutes."
        ),
        "strategies": [
            "Choose one priority.",
            "Include rest breaks in your routine."
        ]
    },
    {
        "id": "RM005",
        "quality": "self_compassion",
        "name": "Self-Compassion Approach",
        "suitable_concerns": [
            "sadness", "low_confidence", "stress"
        ],
        "exercise": (
            "Write one kind and realistic sentence "
            "you would say to a friend."
        ),
        "strategies": [
            "Avoid harsh self-judgment.",
            "Recognize that mistakes are part of learning."
        ]
    },
    {
        "id": "RM006",
        "quality": "problem_solving",
        "name": "Problem-Solving Approach",
        "suitable_concerns": [
            "stress", "anxiety", "anger", "motivation"
        ],
        "exercise": (
            "Write down the problem, two possible solutions, "
            "and one manageable next step."
        ),
        "strategies": [
            "Describe the problem clearly.",
            "Compare possible solutions."
        ]
    }
]

if not KNOWLEDGE_FILE.exists():
    with open(KNOWLEDGE_FILE, "w", encoding="utf-8") as file:
        json.dump(default_knowledge, file, indent=4)

# Load and validate the knowledge base.
with open(KNOWLEDGE_FILE, "r", encoding="utf-8") as file:
    knowledge_base = json.load(file)

if not isinstance(knowledge_base, list) or not knowledge_base:
    raise ValueError("The knowledge base is empty or invalid.")

required_fields = {
    "quality", "name", "suitable_concerns", "exercise"
}

for item in knowledge_base:
    if not required_fields.issubset(item.keys()):
        raise ValueError(
            "A knowledge-base entry is missing required fields."
        )


# ------------------------------------------------------------
# 3. CREATE A DEFAULT PROFILE IF MISSING
# ------------------------------------------------------------

default_profile = {
    "support_style": "empathetic",
    "main_goal": "manage_stress",
    "response_length": "short"
}

if not PROFILE_FILE.exists():
    with open(PROFILE_FILE, "w", encoding="utf-8") as file:
        json.dump(default_profile, file, indent=4)

with open(PROFILE_FILE, "r", encoding="utf-8") as file:
    user_profile = json.load(file)

allowed_styles = ["empathetic", "motivational", "practical"]

if user_profile.get("support_style") not in allowed_styles:
    user_profile["support_style"] = "empathetic"


# ------------------------------------------------------------
# 4. BASIC CONCERN DETECTION
# ------------------------------------------------------------

CONCERN_KEYWORDS = {
    "stress": [
        "stress", "stressed", "overwhelmed", "pressure"
    ],
    "sadness": [
        "sad", "unhappy", "lonely", "heartbroken"
    ],
    "anxiety": [
        "anxious", "anxiety", "worried", "nervous", "panic"
    ],
    "low_confidence": [
        "not confident", "low confidence",
        "not good enough", "self-doubt", "self doubt"
    ],
    "motivation": [
        "unmotivated", "no motivation",
        "procrastinating", "lost motivation"
    ],
    "anger": [
        "angry", "frustrated", "irritated", "furious"
    ]
}

def detect_concern(message):
    text = message.lower()
    detected = []

    for concern, keywords in CONCERN_KEYWORDS.items():
        for keyword in keywords:
            pattern = (
                r"(?<!\w)"
                + re.escape(keyword)
                + r"(?!\w)"
            )

            if re.search(pattern, text):
                detected.append(concern)
                break

    return detected


# ------------------------------------------------------------
# 5. BASIC SAFETY ROUTING
# ------------------------------------------------------------

CRISIS_PHRASES = [
    "kill myself",
    "suicide",
    "end my life",
    "want to die",
    "hurt myself",
    "self harm",
    "self-harm"
]

def safety_check(message):
    text = message.lower()

    matches = [
        phrase for phrase in CRISIS_PHRASES
        if re.search(
            r"(?<!\w)" + re.escape(phrase) + r"(?!\w)",
            text
        )
    ]

    return bool(matches)


# ------------------------------------------------------------
# 6. SELECT A PERSONALIZED STRATEGY
# ------------------------------------------------------------

STYLE_MAPPING = {
    "empathetic": "self_compassion",
    "motivational": "resilience",
    "practical": "problem_solving"
}

GOAL_MAPPING = {
    "manage_stress": "stress",
    "improve_confidence": "low_confidence",
    "build_routines": "motivation",
    "handle_anxiety": "anxiety",
    "general_wellness": None
}

def select_strategy(concern):
    if concern is None:
        return None

    candidates = [
        item for item in knowledge_base
        if concern in item.get("suitable_concerns", [])
    ]

    if not candidates:
        return None

    preferred_quality = STYLE_MAPPING.get(
        user_profile.get("support_style", "empathetic"),
        "self_compassion"
    )

    for item in candidates:
        if item.get("quality") == preferred_quality:
            return item

    return candidates[0]


# ------------------------------------------------------------
# 7. PERSONALIZED AGENT
# ------------------------------------------------------------

def personalized_agent(message):

    if not isinstance(message, str):
        return {
            "status": "invalid_input",
            "response": "Please enter your message as text."
        }

    message = message.strip()

    if not message:
        return {
            "status": "invalid_input",
            "response": "Please enter a message."
        }

    if len(message) > 3000:
        return {
            "status": "invalid_input",
            "response": "Please shorten your message."
        }

    # Safety routing before normal response generation.
    if safety_check(message):
        return {
            "status": "safety_first",
            "strategy": None,
            "response": (
                "I'm sorry you're going through this. Your safety "
                "matters. If you may act on these thoughts or are "
                "in immediate danger, contact local emergency services "
                "and stay with a trusted person if possible. Please "
                "seek qualified professional support. Are you in "
                "immediate danger right now?"
            )
        }

    detected = detect_concern(message)

    concern = detected[0] if detected else GOAL_MAPPING.get(
        user_profile.get("main_goal", "general_wellness")
    )

    if concern is None:
        return {
            "status": "needs_clarification",
            "concern": None,
            "strategy": None,
            "response": (
                "I'm here to listen. What has been on your mind, "
                "and what kind of support would help you?"
            )
        }

    strategy = select_strategy(concern)

    if strategy is None:
        return {
            "status": "needs_clarification",
            "concern": concern,
            "strategy": None,
            "response": (
                "Could you tell me a little more about what feels "
                "most difficult right now?"
            )
        }

    # Use the language model only if it is already available.
    model_ready = (
        callable(globals().get("generate_supportive_response"))
        and "model" in globals()
        and "tokenizer" in globals()
    )

    if model_ready:
        prompt = (
            "Respond empathetically to this user message: "
            + message
            + "\nPossible concern: "
            + concern
            + "\nPreferred style: "
            + user_profile["support_style"]
            + "\nSuggested approach: "
            + strategy["name"]
            + "\nOptional exercise: "
            + strategy["exercise"]
            + "\nAvoid diagnosis or medication advice. "
            "Keep the response concise and ask one gentle question."
        )

        response = generate_supportive_response(prompt)
        source = "language_model"
    else:
        response = (
            "Thank you for sharing that with me. It sounds like "
            + concern.replace("_", " ")
            + " may be part of what you're experiencing. One approach "
            "we could explore is "
            + strategy["name"]
            + ". You could try this optional exercise: "
            + strategy["exercise"]
            + " Would you like to explore this together?"
        )
        source = "template_fallback"

    return {
        "status": "support_response",
        "concern": concern,
        "support_style": user_profile["support_style"],
        "strategy": strategy["name"],
        "exercise": strategy["exercise"],
        "response_source": source,
        "response": response
    }


# ------------------------------------------------------------
# 8. TEST THE AGENT
# ------------------------------------------------------------

print("=" * 60)
print("STEP 11.1: PERSONALIZED AGENT TESTS")
print("=" * 60)

test_messages = [
    "I feel stressed about my exams.",
    "I am not confident in class.",
    "Hello, I need some guidance.",
    "I want to kill myself."
]

for index, message in enumerate(test_messages, start=1):
    result = personalized_agent(message)

    print(f"\nTEST {index}")
    print("User:", message)
    print("Status:", result["status"])
    print("Response:", result["response"])

# ------------------------------------------------------------
# 9. SAVE A NORMAL DEMONSTRATION RESULT
# ------------------------------------------------------------

sample_result = personalized_agent(
    "I feel stressed about my exams."
)

with open(OUTPUT_FILE, "w", encoding="utf-8") as file:
    json.dump(sample_result, file, indent=4, ensure_ascii=False)

# Verification checks
assert sample_result["status"] == "support_response"
assert personalized_agent(
    "I want to kill myself."
)["status"] == "safety_first"
assert OUTPUT_FILE.exists()

print("\nKnowledge base entries:", len(knowledge_base))
print("Profile loaded: PASSED")
print("Normal response test: PASSED")
print("Safety routing test: PASSED")
print("Saved result:", OUTPUT_FILE)
print("STEP 11.1 SUCCESSFULLY COMPLETED")
```

```python

# ============================================================
# PROJECT 1: HUMAN ROLE-MODEL INSPIRED COUNSELLING AGENT
# STEP 12: SAFETY AND CRISIS-RESPONSE HANDLING
# ============================================================

from pathlib import Path
import json
import re

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

SAFETY_DIR = PROJECT_DIR / "src" / "safety"
OUTPUT_DIR = PROJECT_DIR / "outputs"

SAFETY_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# ------------------------------------------------------------
# 1. Define basic safety indicators
# ------------------------------------------------------------

URGENT_PHRASES = [
    "kill myself",
    "suicide",
    "end my life",
    "want to die",
    "hurt myself",
    "self harm",
    "self-harm",
    "take my own life",
    "better off dead"
]

# These are illustrative indicators, not a complete
# mental-health risk assessment.


# ------------------------------------------------------------
# 2. Safety screening function
# ------------------------------------------------------------

def screen_message(message):
    if not isinstance(message, str):
        return {
            "valid": False,
            "route": "INVALID_INPUT",
            "reason": "Input must be text."
        }

    message = message.strip()

    if not message:
        return {
            "valid": False,
            "route": "INVALID_INPUT",
            "reason": "The message is empty."
        }

    if len(message) > 5000:
        return {
            "valid": False,
            "route": "INVALID_INPUT",
            "reason": "The message exceeds 5000 characters."
        }

    normalized = message.lower()

    matches = [
        phrase for phrase in URGENT_PHRASES
        if re.search(
            r"(?<!\w)" + re.escape(phrase) + r"(?!\w)",
            normalized
        )
    ]

    if matches:
        return {
            "valid": True,
            "route": "SAFETY_FIRST",
            "reason": "Potential urgent safety concern detected.",
            "matched_indicators": matches
        }

    return {
        "valid": True,
        "route": "NORMAL_SUPPORT",
        "reason": "No configured urgent phrase detected.",
        "matched_indicators": []
    }


# ------------------------------------------------------------
# 3. Create a safety-first response
# ------------------------------------------------------------

def safety_first_response():
    return (
        "I'm sorry you're going through such a difficult time. "
        "I'm glad you shared this. Your safety matters. "
        "If you might act on these thoughts or are in immediate "
        "danger, contact local emergency services now and try to "
        "stay with a trusted person who can support you. "
        "Please consider contacting a qualified mental health "
        "professional or crisis service. "
        "Are you in immediate danger right now?"
    )


# ------------------------------------------------------------
# 4. Create a safe routing function
# ------------------------------------------------------------

def safety_router(message):
    result = screen_message(message)

    if not result["valid"]:
        return {
            "status": "invalid_input",
            "safety_check": result,
            "response": result["reason"]
        }

    if result["route"] == "SAFETY_FIRST":
        return {
            "status": "safety_first",
            "safety_check": result,
            "normal_counselling_allowed": False,
            "response": safety_first_response()
        }

    return {
        "status": "normal_support",
        "safety_check": result,
        "normal_counselling_allowed": True,
        "response": (
            "Thank you for sharing. Could you tell me a little "
            "more about what kind of support would help you?"
        )
    }


# ------------------------------------------------------------
# 5. Test the safety module
# ------------------------------------------------------------

test_cases = [
    ("I feel stressed about exams.", "normal_support"),
    ("I want to kill myself.", "safety_first"),
    ("I would like someone to talk to.", "normal_support"),
    ("", "invalid_input"),
    (None, "invalid_input"),
    ("I want to end my life.", "safety_first")
]

print("=" * 65)
print("STEP 12: SAFETY ROUTING TESTS")
print("=" * 65)

test_results = []

for index, (message, expected_status) in enumerate(
    test_cases, start=1
):
    result = safety_router(message)

    passed = result["status"] == expected_status

    test_results.append({
        "test_number": index,
        "expected": expected_status,
        "actual": result["status"],
        "passed": passed
    })

    print(f"\nTest {index}")
    print("Input:", repr(message))
    print("Expected:", expected_status)
    print("Actual:", result["status"])
    print("Result:", "PASSED" if passed else "FAILED")

# ------------------------------------------------------------
# 6. Save the reusable safety module
# ------------------------------------------------------------

module_source = '''
import re

URGENT_PHRASES = [
    "kill myself", "suicide", "end my life",
    "want to die", "hurt myself", "self harm",
    "self-harm", "take my own life", "better off dead"
]

def screen_message(message):
    if not isinstance(message, str):
        return {"valid": False, "route": "INVALID_INPUT",
                "reason": "Input must be text."}

    message = message.strip()

    if not message:
        return {"valid": False, "route": "INVALID_INPUT",
                "reason": "The message is empty."}

    if len(message) > 5000:
        return {"valid": False, "route": "INVALID_INPUT",
                "reason": "The message exceeds 5000 characters."}

    normalized = message.lower()
    matches = [
        phrase for phrase in URGENT_PHRASES
        if re.search(
            r"(?<!\\\\w)" + re.escape(phrase) + r"(?!\\\\w)",
            normalized
        )
    ]

    return {
        "valid": True,
        "route": "SAFETY_FIRST" if matches else "NORMAL_SUPPORT",
        "reason": (
            "Potential urgent safety concern detected."
            if matches else "No configured urgent phrase detected."
        ),
        "matched_indicators": matches
    }
'''

module_path = SAFETY_DIR / "safety_router.py"
module_path.write_text(
    module_source.strip() + "\n",
    encoding="utf-8"
)

# ------------------------------------------------------------
# 7. Verify and save test results
# ------------------------------------------------------------

results_file = OUTPUT_DIR / "safety_test_results.json"

with open(results_file, "w", encoding="utf-8") as file:
    json.dump(test_results, file, indent=4)

assert all(item["passed"] for item in test_results)
assert module_path.exists()
assert results_file.exists()

print("\n" + "=" * 65)
print("Passed tests:", sum(item["passed"] for item in test_results))
print("Total tests:", len(test_results))
print("Safety module saved:", module_path)
print("Test results saved:", results_file)
print("STEP 12 SUCCESSFULLY COMPLETED")
print("=" * 65)
```

```python

# ============================================================
# STEP 13: TESTING AND EVALUATION
# ============================================================

from pathlib import Path
import json
import re

PROJECT_DIR = Path(
    "/content/Human_Role_Model_Counselling_Agent"
)

OUTPUT_DIR = PROJECT_DIR / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 1. Define test categories
# ------------------------------------------------------------

KEYWORDS = {
    "stress": ["stress", "stressed", "overwhelmed", "pressure"],
    "sadness": ["sad", "unhappy", "lonely"],
    "anxiety": ["anxious", "anxiety", "worried", "nervous"],
    "low_confidence": ["not confident", "low confidence"],
    "motivation": ["no motivation", "unmotivated", "procrastinating"],
    "anger": ["angry", "frustrated", "irritated"]
}

CRISIS_PHRASES = [
    "kill myself", "suicide", "end my life",
    "want to die", "hurt myself", "self harm", "self-harm"
]

STRATEGIES = {
    "stress": {
        "name": "Resilience Approach",
        "exercise": "Choose one small, manageable task."
    },
    "sadness": {
        "name": "Self-Compassion Approach",
        "exercise": "Write one kind sentence to yourself."
    },
    "anxiety": {
        "name": "Mindfulness Approach",
        "exercise": "Notice five things you can see."
    },
    "low_confidence": {
        "name": "Self-Confidence Approach",
        "exercise": "Identify one personal strength."
    },
    "motivation": {
        "name": "Healthy Routine Approach",
        "exercise": "Work on one task for 15 minutes."
    },
    "anger": {
        "name": "Problem-Solving Approach",
        "exercise": "Pause and identify one safe next step."
    }
}


# ------------------------------------------------------------
# 2. Create a testable evaluation pipeline
# ------------------------------------------------------------

def evaluate_agent(message):
    if not isinstance(message, str) or not message.strip():
        return {
            "status": "invalid_input",
            "concern": None,
            "strategy": None
        }

    message = message.strip()

    if len(message) > 5000:
        return {
            "status": "invalid_input",
            "concern": None,
            "strategy": None
        }

    text = message.lower()

    # Safety screening comes first.
    crisis_detected = any(
        re.search(
            r"(?<!\w)" + re.escape(phrase) + r"(?!\w)",
            text
        )
        for phrase in CRISIS_PHRASES
    )

    if crisis_detected:
        return {
            "status": "safety_first",
            "concern": None,
            "strategy": None
        }

    # Detect possible concerns.
    detected = []

    for concern, words in KEYWORDS.items():
        for word in words:
            pattern = (
                r"(?<!\w)" + re.escape(word) + r"(?!\w)"
            )

            if re.search(pattern, text):
                detected.append(concern)
                break

    if not detected:
        return {
            "status": "needs_clarification",
            "concern": None,
            "strategy": None
        }

    concern = detected[0]

    return {
        "status": "support_response",
        "concern": concern,
        "strategy": STRATEGIES[concern]["name"],
        "exercise": STRATEGIES[concern]["exercise"]
    }


# ------------------------------------------------------------
# 3. Prepare expected test results
# ------------------------------------------------------------

test_cases = [
    {
        "message": "I am stressed about my exams.",
        "expected": "support_response"
    },
    {
        "message": "I feel sad and lonely.",
        "expected": "support_response"
    },
    {
        "message": "I am worried about my future.",
        "expected": "support_response"
    },
    {
        "message": "I am not confident in class.",
        "expected": "support_response"
    },
    {
        "message": "I have no motivation to study.",
        "expected": "support_response"
    },
    {
        "message": "I feel angry and frustrated.",
        "expected": "support_response"
    },
    {
        "message": "Hello, how are you?",
        "expected": "needs_clarification"
    },
    {
        "message": "",
        "expected": "invalid_input"
    },
    {
        "message": "I want to kill myself.",
        "expected": "safety_first"
    },
    {
        "message": "I want to end my life.",
        "expected": "safety_first"
    }
]


# ------------------------------------------------------------
# 4. Execute the tests
# ------------------------------------------------------------

results = []

print("=" * 65)
print("STEP 13: AGENT EVALUATION")
print("=" * 65)

for number, case in enumerate(test_cases, start=1):
    actual = evaluate_agent(case["message"])
    passed = actual["status"] == case["expected"]

    record = {
        "test_number": number,
        "message": case["message"],
        "expected_status": case["expected"],
        "actual_status": actual["status"],
        "passed": passed,
        "detected_concern": actual["concern"],
        "selected_strategy": actual["strategy"]
    }

    results.append(record)

    print(
        f"Test {number}: "
        f"{'PASSED' if passed else 'FAILED'}"
    )
    print("  Expected:", case["expected"])
    print("  Actual:", actual["status"])

# ------------------------------------------------------------
# 5. Calculate evaluation metrics
# ------------------------------------------------------------

total_tests = len(results)
passed_tests = sum(item["passed"] for item in results)
failed_tests = total_tests - passed_tests

accuracy = (
    passed_tests / total_tests
) * 100 if total_tests else 0.0

summary = {
    "total_tests": total_tests,
    "passed_tests": passed_tests,
    "failed_tests": failed_tests,
    "test_status_accuracy_percent": round(accuracy, 2),
    "interpretation": (
        "Accuracy measures agreement with these predefined "
        "test expectations only. It is not clinical accuracy "
        "or proof of crisis-detection reliability."
    ),
    "results": results
}

# ------------------------------------------------------------
# 6. Save evaluation results
# ------------------------------------------------------------

result_file = OUTPUT_DIR / "evaluation_results.json"

with open(result_file, "w", encoding="utf-8") as file:
    json.dump(summary, file, indent=4, ensure_ascii=False)

# ------------------------------------------------------------
# 7. Print summary
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("EVALUATION SUMMARY")
print("=" * 65)

print("Total tests:", total_tests)
print("Passed:", passed_tests)
print("Failed:", failed_tests)
print("Test-status accuracy:", f"{accuracy:.2f}%")
print("Results saved to:", result_file)

if failed_tests == 0:
    print("STEP 13 SUCCESSFULLY COMPLETED")
else:
    print("Some tests failed. Review the results before proceeding.")

print("=" * 65)
```

```python

# ============================================================
# MINDFUL MENTOR
# Human Role-Model Inspired Personalized Mental Health
# Counselling Agent
# Complete website with 20 dropdown prompts
# Python + Gradio
# ============================================================

%pip install -q gradio

import gradio as gr
import re
import random

# ============================================================
# 1. COUNSELLING STRATEGIES
# ============================================================

STRATEGIES = {
    "stress": (
        "Resilience",
        "Choose one small task and focus on the next manageable step."
    ),
    "anxiety": (
        "Mindfulness",
        "Notice five things you can see around you, if comfortable."
    ),
    "sadness": (
        "Self-Compassion",
        "Think about what kind words you would offer a friend."
    ),
    "confidence": (
        "Self-Confidence",
        "Write down one personal strength or recent achievement."
    ),
    "motivation": (
        "Healthy Routine",
        "Choose one task and work on it for 10 minutes."
    ),
    "anger": (
        "Problem-Solving",
        "Pause if safe and consider one constructive next step."
    ),
    "time_management": (
        "Time Management",
        "Choose three priorities and schedule realistic time for each."
    ),
    "sleep": (
        "Healthy Sleep Routine",
        "Try a consistent bedtime and a relaxing wind-down routine."
    ),
    "relationships": (
        "Healthy Communication",
        "Consider expressing your feelings calmly and respectfully."
    ),
    "concentration": (
        "Focused Learning",
        "Try one short study session with fewer distractions."
    )
}

KEYWORDS = {
    "stress": [
        "stress", "stressed", "overwhelmed", "pressure",
        "exam stress", "exam stressed", "college work"
    ],
    "anxiety": [
        "anxious", "anxiety", "worried", "nervous", "panic",
        "afraid of failure", "fear of failure"
    ],
    "sadness": [
        "sad", "unhappy", "lonely", "heartbroken"
    ],
    "confidence": [
        "not confident", "low confidence", "self-doubt",
        "self doubt", "confidence", "presentations",
        "speaking in class", "criticism"
    ],
    "motivation": [
        "no motivation", "unmotivated", "procrastinating",
        "lost motivation", "better habits", "resilient"
    ],
    "anger": [
        "angry", "frustrated", "irritated", "furious"
    ],
    "time_management": [
        "manage my time", "time management", "balance study",
        "study and relaxation", "schedule"
    ],
    "sleep": [
        "sleep", "sleeping", "bedtime"
    ],
    "relationships": [
        "friendship", "friendships", "relationship",
        "friend", "communication"
    ],
    "concentration": [
        "concentration", "concentrate", "focus",
        "mindfulness", "healthy routine", "study routine"
    ]
}

CRISIS_PHRASES = [
    "kill myself",
    "suicide",
    "end my life",
    "want to die",
    "hurt myself",
    "self harm",
    "self-harm",
    "take my own life",
    "better off dead"
]

AFFIRMATIONS = [
    "💙 Small progress is still progress.",
    "💙 I can take one step at a time.",
    "💙 I deserve patience and kindness.",
    "💙 I can learn from challenges.",
    "💙 I do not need to solve everything today.",
    "💙 Asking for support is a strength."
]

# ============================================================
# 2. HELPER FUNCTION
# ============================================================

def contains_phrase(text, phrase):
    pattern = r"(?<!\w)" + re.escape(phrase) + r"(?!\w)"
    return bool(re.search(pattern, text.lower()))


# ============================================================
# 3. COUNSELLING RESPONSE ENGINE
# ============================================================

def create_response(message):

    if not isinstance(message, str) or not message.strip():
        return (
            "### 💙 Please enter a message\n\n"
            "Type a message or select a quick prompt."
        )

    message = message.strip()

    if len(message) > 3000:
        return (
            "### Message too long\n\n"
            "Please shorten your message to 3000 characters or fewer."
        )

    # Basic safety-first routing
    if any(
        contains_phrase(message, phrase)
        for phrase in CRISIS_PHRASES
    ):
        return (
            "### 💙 Your safety matters\n\n"
            "I'm sorry you're going through such a difficult time. "
            "If you may act on these thoughts or are in immediate "
            "danger, contact local emergency services and try to stay "
            "with a trusted person. Please seek qualified professional "
            "support.\n\n"
            "Are you in immediate danger right now?"
        )

    # Special case for affirmation prompt
    if contains_phrase(message, "positive affirmation"):
        return (
            "### ✨ A positive reminder\n\n"
            + random.choice(AFFIRMATIONS)
            + "\n\nTake what feels helpful and leave the rest."
        )

    # Detect a possible concern
    concern = None

    for category, words in KEYWORDS.items():
        if any(
            contains_phrase(message, word)
            for word in words
        ):
            concern = category
            break

    # General conversation
    if concern is None:
        return (
            "### 💙 Welcome to Mindful Mentor\n\n"
            "Thank you for reaching out. I'm here to listen "
            "without judgment.\n\n"
            "Could you tell me a little more about what has "
            "been on your mind?"
        )

    strategy_name, exercise = STRATEGIES[concern]

    return (
        "### 💙 Thank you for sharing\n\n"
        f"**Possible concern:** {concern.replace('_', ' ').title()}\n\n"
        f"**Suggested approach:** {strategy_name}\n\n"
        f"**Optional exercise:** {exercise}\n\n"
        "Would you like to tell me a little more about your situation?"
    )


# ============================================================
# 4. MESSAGE CALLBACKS
# ============================================================

def send_message(message):

    try:
        answer = create_response(message)
        return answer, ""

    except Exception as error:
        print("Response error:", repr(error))

        return (
            "### 💙 Unable to process this message\n\n"
            "Please try again with a shorter message.",
            message or ""
        )


def clear_response():
    return (
        "### 💙 Welcome to Mindful Mentor\n\n"
        "Your supportive response will appear here.",
        ""
    )


# ============================================================
# 5. MOOD CHECK-IN
# ============================================================

def check_mood(mood):

    suggestions = {
        "Happy": "Enjoy the moment and notice what is going well.",
        "Calm": "Continue at a comfortable pace.",
        "Okay": "Choose one small, meaningful task today.",
        "Stressed": "Take a short break and identify one priority.",
        "Anxious": (
            "Notice your surroundings and consider talking "
            "to someone you trust."
        ),
        "Sad": (
            "Be kind to yourself and consider reaching "
            "out to someone you trust."
        ),
        "Angry": (
            "If safe, take some space before deciding how to respond."
        )
    }

    return (
        f"### Your mood: {mood}\n\n"
        f"{suggestions.get(mood, 'Take a moment for yourself.')}\n\n"
        "This is a reflection aid, not a diagnosis."
    )


# ============================================================
# 6. WELLNESS FEATURES
# ============================================================

def breathing_exercise():

    return (
        "### 🌿 Gentle breathing break\n\n"
        "1. Sit comfortably, if possible.\n\n"
        "2. Breathe in gently for about 4 seconds.\n\n"
        "3. Breathe out slowly for about 6 seconds.\n\n"
        "4. Repeat at a comfortable pace.\n\n"
        "**Remember:** Do not force your breathing. Stop if you "
        "feel uncomfortable or dizzy."
    )


def positive_affirmation():

    return (
        "### ✨ Your positive reminder\n\n"
        + random.choice(AFFIRMATIONS)
        + "\n\nTake what feels helpful and leave the rest."
    )


# ============================================================
# 7. ALL 20 QUICK PROMPTS
# ============================================================

QUICK_PROMPTS = [
    "I feel stressed about my exams.",
    "I lack confidence when speaking in class.",
    "I have no motivation to study.",
    "I feel anxious about my future.",
    "I feel sad and need support.",
    "Help me manage my time better.",
    "I am procrastinating on assignments.",
    "I am afraid of failure.",
    "I feel overwhelmed by college work.",
    "How can I improve my concentration?",
    "I have difficulty sleeping before exams.",
    "How can I develop a healthy routine?",
    "I get nervous during presentations.",
    "How can I handle criticism?",
    "I want to build better habits.",
    "How can I practise mindfulness?",
    "I am struggling with a friendship.",
    "How can I balance study and relaxation?",
    "I want to become more resilient.",
    "Give me a positive affirmation."
]


# ============================================================
# 8. BLUE WEBSITE CSS
# ============================================================

CSS = """
.gradio-container {
    background: linear-gradient(
        135deg,
        #061a40 0%,
        #0755a5 55%,
        #1598c8 100%
    ) !important;
}

/* Header */
#header {
    text-align: center;
    background: linear-gradient(120deg, #082b63, #087eb8);
    border: 1px solid #63c7ff;
    border-radius: 20px;
    padding: 24px;
    margin-bottom: 16px;
}

#header h1 {
    color: #ffffff !important;
    font-size: 34px !important;
}

#header p {
    color: #e2f5ff !important;
}

/* Agent response panel */
#response-panel {
    background: #dbeafe !important;
    color: #082b63 !important;
    border: 2px solid #60a5fa !important;
    border-radius: 18px !important;
    padding: 20px !important;
}

/* Readable dark-blue response text */
#response-panel,
#response-panel p,
#response-panel strong,
#response-panel em,
#response-panel h1,
#response-panel h2,
#response-panel h3,
#response-panel li,
#response-panel span {
    color: #082b63 !important;
}

/* Prompt dropdown */
#prompt-dropdown {
    border: 1px solid #60a5fa !important;
    border-radius: 12px !important;
}

/* Buttons */
button {
    border-radius: 10px !important;
}

/* Footer */
footer {
    display: none !important;
}
"""


# ============================================================
# 9. BUILD WEBSITE
# ============================================================

with gr.Blocks(
    css=CSS,
    title="Mindful Mentor - AI Wellness Companion",
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="sky"
    )
) as demo:

    gr.HTML("""
    <div id="header">
        <h1>💙 Mindful Mentor</h1>
        <p>Human Role-Model Inspired Personalized Wellness Companion</p>
        <p>Resilience • Mindfulness • Confidence • Self-Compassion</p>
    </div>
    """)

    with gr.Row():

        # LEFT COLUMN
        with gr.Column(scale=2):

            gr.Markdown("## 💬 AI Wellness Chat")

            message = gr.Textbox(
                label="Your message",
                placeholder="Example: I feel stressed about my exams.",
                lines=3,
                max_lines=5
            )

            with gr.Row():
                send = gr.Button(
                    "Send Message ➤",
                    variant="primary"
                )

                clear = gr.Button("Clear")

            # Only one dropdown instead of 20 visible buttons
            gr.Markdown("### ⚡ Quick Prompts")

            prompt_dropdown = gr.Dropdown(
                choices=QUICK_PROMPTS,
                value=QUICK_PROMPTS[0],
                label="Choose from 20 testing scenarios",
                interactive=True,
                elem_id="prompt-dropdown"
            )

            use_prompt = gr.Button(
                "Use Selected Prompt",
                variant="primary"
            )

            gr.Markdown("### 💬 Agent Response")

            response = gr.Markdown(
                value=(
                    "### 💙 Welcome to Mindful Mentor\n\n"
                    "Your supportive response will appear here."
                ),
                elem_id="response-panel"
            )

        # RIGHT COLUMN
        with gr.Column(scale=1):

            gr.Markdown("## 💗 Mood Check-in")

            mood = gr.Dropdown(
                choices=[
                    "Happy",
                    "Calm",
                    "Okay",
                    "Stressed",
                    "Anxious",
                    "Sad",
                    "Angry"
                ],
                value="Okay",
                label="How are you feeling?"
            )

            mood_button = gr.Button("Check My Mood")

            mood_output = gr.Markdown(
                "Your mood suggestion will appear here."
            )

            gr.Markdown("## 🌿 Relaxation Corner")

            breathing_button = gr.Button(
                "Breathing Exercise"
            )

            breathing_output = gr.Markdown(
                "Take a comfortable break whenever you need one."
            )

            affirmation_button = gr.Button(
                "✨ Positive Affirmation"
            )

            affirmation_output = gr.Markdown(
                "Your positive reminder will appear here."
            )

    gr.Markdown(
        "---\n"
        "**Academic prototype:** This application does not diagnose "
        "mental health conditions or replace professional care. "
        "Avoid entering identifying or sensitive personal information."
    )

    # ========================================================
    # 10. ALL EVENT CONNECTIONS INSIDE BLOCKS
    # ========================================================

    send.click(
        fn=send_message,
        inputs=[message],
        outputs=[response, message]
    )

    message.submit(
        fn=send_message,
        inputs=[message],
        outputs=[response, message]
    )

    clear.click(
        fn=clear_response,
        inputs=[],
        outputs=[response, message]
    )

    # Select a prompt and transfer it to the message input
    use_prompt.click(
        fn=lambda selected: selected or "",
        inputs=[prompt_dropdown],
        outputs=[message]
    )

    mood_button.click(
        fn=check_mood,
        inputs=[mood],
        outputs=[mood_output]
    )

    breathing_button.click(
        fn=breathing_exercise,
        inputs=[],
        outputs=[breathing_output]
    )

    affirmation_button.click(
        fn=positive_affirmation,
        inputs=[],
        outputs=[affirmation_output]
    )


# ============================================================
# 11. LAUNCH PUBLIC WEBSITE
# ============================================================

print("Starting Mindful Mentor website...")
print("Wait for the public Gradio URL.")

demo.launch(
    share=True,
    debug=False,
    inbrowser=False
)
```
